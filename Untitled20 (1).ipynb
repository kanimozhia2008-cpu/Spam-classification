```python
# STEP 21: Check ID Columns

print("Sample nameOrig values:")
print(df["nameOrig"].head())

print("\nSample nameDest values:")
print(df["nameDest"].head())

print("\nUnique values:")
print("nameOrig:", df["nameOrig"].nunique())
print("nameDest:", df["nameDest"].nunique())
```

```python
# STEP 22: Drop ID Columns

df = df.drop(["nameOrig", "nameDest"], axis=1)

print("Remaining columns:")
print(df.columns.tolist())
```

```python
# STEP 23: Encode Transaction Type

df = pd.get_dummies(df, columns=["type"], dtype=int)

print("Columns after encoding:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python
# STEP 24: Separate Features and Target

X = df.drop("isFraud", axis=1)
y = df["isFraud"]

print("Features (X) shape:", X.shape)
print("Target (y) shape:", y.shape)

print("\nFeature columns:")
print(X.columns.tolist())
```

```python
# STEP 25: Train-Test Split

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
# STEP 26: Feature Scaling

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("X_train_scaled shape:", X_train_scaled.shape)
print("X_test_scaled shape:", X_test_scaled.shape)
```

```python
# STEP 27: Train Logistic Regression

from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)

model.fit(X_train_scaled, y_train)

print("Logistic Regression model trained successfully!")
```

```python
# STEP 28: Make Predictions

y_pred = model.predict(X_test_scaled)

print("First 20 Predictions:")
print(y_pred[:20])
```

```python
# STEP 29: Model Evaluation

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1 Score :", f1)
```

```python
# STEP 30: Confusion Matrix

from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Not Fraud", "Fraud"]
)

disp.plot()
plt.title("Confusion Matrix - Fraud Detection")
plt.show()
```

```python
# STEP 31: Classification Report

from sklearn.metrics import classification_report

print("Classification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["Not Fraud", "Fraud"]
))
```

```python
# STEP 32: ROC Curve and AUC

from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

# Fraud probability
y_prob = model.predict_proba(X_test_scaled)[:, 1]

# ROC values
fpr, tpr, thresholds = roc_curve(y_test, y_prob)

# AUC
auc = roc_auc_score(y_test, y_prob)

print("AUC Score:", auc)

# Plot ROC Curve
plt.figure(figsize=(8, 6))

plt.plot(fpr, tpr, label=f"AUC = {auc:.4f}")
plt.plot([0, 1], [0, 1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Fraud Detection")
plt.legend()

plt.show()
```

```python
# STEP 33: User Input Prediction

import pandas as pd

print("Enter Transaction Details")

step = float(input("Step: "))
amount = float(input("Amount: "))
oldbalanceOrg = float(input("Old Balance Origin: "))
newbalanceOrig = float(input("New Balance Origin: "))
oldbalanceDest = float(input("Old Balance Destination: "))
newbalanceDest = float(input("New Balance Destination: "))
isFlaggedFraud = int(input("Is Flagged Fraud (0/1): "))

print("\nTransaction Type:")
print("1 - CASH_IN")
print("2 - CASH_OUT")
print("3 - DEBIT")
print("4 - PAYMENT")
print("5 - TRANSFER")

transaction_type = int(input("Enter type number: "))

# Create input with same feature structure as X
new_data = pd.DataFrame(0, index=[0], columns=X.columns)

new_data["step"] = step
new_data["amount"] = amount
new_data["oldbalanceOrg"] = oldbalanceOrg
new_data["newbalanceOrig"] = newbalanceOrig
new_data["oldbalanceDest"] = oldbalanceDest
new_data["newbalanceDest"] = newbalanceDest
new_data["isFlaggedFraud"] = isFlaggedFraud

type_columns = {
    1: "type_CASH_IN",
    2: "type_CASH_OUT",
    3: "type_DEBIT",
    4: "type_PAYMENT",
    5: "type_TRANSFER"
}

new_data[type_columns[transaction_type]] = 1

# Scale the new input
new_data_scaled = scaler.transform(new_data)

# Prediction
prediction = model.predict(new_data_scaled)[0]
probability = model.predict_proba(new_data_scaled)[0][1]

if prediction == 1:
    print("\n🚨 Prediction: FRAUD")
else:
    print("\n✅ Prediction: NOT FRAUD")

print(f"Fraud Probability: {probability:.2%}")
```
